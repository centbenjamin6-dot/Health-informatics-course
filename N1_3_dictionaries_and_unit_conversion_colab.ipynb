## Part A · Dictionaries: labelled records
📘 A list finds items by **position** (`weights[2]`). A **dictionary** finds items by **label**, called a **key**. This suits a patient record: you ask for `"age"`, not for "the fourth thing".
```
patient = {"id": "P001", "age": 54}
│         │ │     │  │      │
│         │ │     │  │      └── value
│         │ │     │  └── colon between key and value
│         │ │     └── the key is usually text, so it sits in quotes
│         │ └── curly braces { } mark a dictionary
│         └── each "key: value" pair is separated from the next by a comma
└── the variable name
```
| Task | Syntax | Example |
|---|---|---|
| read a value | `record["key"]` | `patient["age"]` → `54` |
| change a value | `record["key"] = new` | `patient["age"] = 55` |
| add a new field | `record["new_key"] = value` | `patient["smoker"] = False` |
| safe read (no crash if missing) | `record.get("key", default)` | `patient.get("smoker", "unknown")` |
| all keys / values | `list(record.keys())`, `list(record.values())` | |
| does a key exist? | `"age" in record` | `True` |

⚠️ **Square brackets `[ ]` for looking up, curly braces `{ }` for creating.**

### ▶️ Worked example A1: one patient record

````python
patient = {
    "id": "P001",
    "age": 54,
    "sex": "F",
    "weight_kg": 68.0,
    "height_cm": 165,
    "glucose_mgdl": 126,          # fasting glucose in mg/dL
    "creatinine_umol_l": 88,      # serum creatinine in µmol/L
}

print(patient["id"])
print(patient["age"])
print(list(patient.keys()))       # what fields exist?
print("age" in patient)           # True
````

### ▶️ Worked example A2: changing, adding and reading safely
The dictionary `visit` records a follow-up visit.

````python
visit = {"date": "2026-03-01", "weight_kg": 68.0}

visit["weight_kg"] = 67.2          # change an existing value
visit["notes"] = "feeling well"    # add a new key
print(visit)

print(visit.get("bp_systolic", "not recorded"))   # key missing -> the default is returned, no crash
````

📘 **What if you ask for a key that does not exist?** Python raises a `KeyError`. Here it is on purpose, caught safely so the notebook keeps running:

````python
try:
    print(visit["bp_systolic"])
except KeyError as error:
    print("KeyError: the key", error, "does not exist. Use .get() to avoid this.")
````

## Part C · Tuples: values that never change
📘 A **tuple** looks like a list but uses round brackets and **cannot be changed** after it is created. Ideal for fixed values such as a **reference range**.

A tuple can be **unpacked** into separate variables in one line:
```
low, high = (70, 99)
│    │       └── a tuple: (minimum, maximum)
│    └── second variable gets 99
└── first variable gets 70
```
### ▶️ Worked example C1: a reference range

````python
hba1c_range = (4.0, 5.6)                 # normal HbA1c (%), illustrative
low, high = hba1c_range                  # unpack
print("Normal range:", low, "to", high)
print(hba1c_range[0])                    # indexing works as for lists

try:
    hba1c_range[0] = 3.0                 # tuples cannot be changed
except TypeError as error:
    print("TypeError:", error)
````

## Part D · Unit conversion
📘 Laboratories and countries report the same measurement in **different units**. Mixing units is a serious real-world error, so we convert carefully and keep conversion factors in **named variables**. By convention, names of fixed values are written in CAPITALS.

| Measurement | From → To | Rule |
|---|---|---|
| Glucose | mg/dL → mmol/L | divide by **18.016** |
| Creatinine | µmol/L → mg/dL | divide by **88.4** |
| Temperature | °C → °F | × 9/5 + 32 |
| Weight | kg → lb | × **2.20462** |
| Height | cm → m | divide by **100** |

*(Educational conversions. In real clinical work always follow your laboratory's stated units and reference intervals.)*

### ▶️ Worked example D1: converting patient P001

````python
GLUCOSE_FACTOR = 18.016        # mg/dL divided by this = mmol/L
CREATININE_FACTOR = 88.4       # µmol/L divided by this = mg/dL

glucose_mgdl = patient["glucose_mgdl"]
glucose_mmol_example = glucose_mgdl / GLUCOSE_FACTOR

print(f"Glucose: {glucose_mgdl} mg/dL = {glucose_mmol_example:.2f} mmol/L")
````

## Part E · Mini-project: patient summary script
📘 **Goal:** for each patient, print **one line** with BMI, body surface area, glucose in mmol/L and creatinine in mg/dL:

`P001 | BMI 25.0 | BSA 1.77 m² | Glucose 7.0 mmol/L | Creatinine 1.00 mg/dL`

Everything you need was taught today. A short reminder of two new tricks:
- `p = patients[0]` makes **`p` a shortcut name** for the first patient's record, so we can write `p["age"]` instead of `patients[0]["age"]`.
- Inside an f-string that uses double quotes, write dictionary keys with **single quotes**: `f"{p['id']}"`.

### ▶️ Worked example E1: patient 1 (index 0)

````python
GLUCOSE_FACTOR = 18.016
CREATININE_FACTOR = 88.4

p = patients[0]                                              # shortcut for the first patient
height_m = p["height_cm"] / 100                              # cm -> m
bmi = p["weight_kg"] / height_m ** 2                         # BMI
bsa = (p["height_cm"] * p["weight_kg"] / 3600) ** 0.5        # Mosteller BSA
glucose_mmol = p["glucose_mgdl"] / GLUCOSE_FACTOR            # mg/dL -> mmol/L
creatinine_mgdl = p["creatinine_umol_l"] / CREATININE_FACTOR # µmol/L -> mg/dL

summary_p1 = f"{p['id']} | BMI {bmi:.1f} | BSA {bsa:.2f} m² | Glucose {glucose_mmol:.1f} mmol/L | Creatinine {creatinine_mgdl:.2f} mg/dL"
print(summary_p1)
````

## 🧾 Recap: your cheat sheet
| Idea | Syntax | Example |
|---|---|---|
| Create a dictionary | `{"key": value, ...}` | `{"age": 54}` |
| Read / change / add | `d["key"]`, `d["key"] = x` | `patient["age"] = 55` |
| Safe read | `d.get("key", default)` | `visit.get("bp", "n/a")` |
| List of records | `[d1, d2, d3]` | `patients[1]["sex"]` |
| Tuple | `(a, b)` | `low, high = (70, 99)` |
| Named constant | `NAME = value` | `GLUCOSE_FACTOR = 18.016` |

### Common slips
- **`KeyError`:** the key is missing or misspelt (`"Age"` is not `"age"`).
- Using `( )` or `{ }` when you need `[ ]` to look something up.
- Forgetting quotes around text keys.
- **Unit mix-ups:** always write the unit in variable names (`glucose_mgdl`, `glucose_mmol`).
- Using double quotes for keys inside a double-quoted f-string.
