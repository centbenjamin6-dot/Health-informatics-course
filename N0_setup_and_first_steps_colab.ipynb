### 🧰 The answer checker
📘 Later in this notebook, cells marked 🎯 **Check** mark your answers with ✅ or ❌. They need a small helper, which the cell below creates. **Run it once now.** You do not need to understand it yet.

````python
# 🧰 COURSE HELPER: just run this cell once. (You will understand every line by Day 2.)
_results = {}

def _report(name, ok, message):
    _results[name] = ok
    print(("✅ " if ok else "❌ ") + message)

def check(name, expected, tol=0, hint=""):
    """Is the variable called `name` equal to the expected answer?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. Fill in the blank or write your code, then run the cell again. {hint}".strip())
        return
    value = globals()[name]
    try:
        if isinstance(expected, bool):
            ok = isinstance(value, bool) and value == expected
        elif isinstance(expected, (int, float)):
            ok = (not isinstance(value, bool)) and abs(value - expected) <= tol
        else:
            ok = value == expected
    except Exception:
        ok = False
    if ok:
        _report(name, True, f"{name}: correct!")
    else:
        _report(name, False, f"{name}: not quite yet. Your value is {value!r}. {hint}".strip())

def check_type(name, expected_type, hint=""):
    """Does the variable called `name` hold the expected type of value?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. {hint}".strip())
        return
    value = globals()[name]
    ok = type(value) is expected_type
    if ok:
        _report(name, True, f"{name}: correct type ({expected_type.__name__})!")
    else:
        _report(name, False, f"{name}: expected type {expected_type.__name__}, but yours is {type(value).__name__}. {hint}".strip())

def check_that(name, test, hint=""):
    """Does the value in `name` pass a custom test?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. {hint}".strip())
        return
    value = globals()[name]
    try:
        ok = bool(test(value))
    except Exception:
        ok = False
    if ok:
        _report(name, True, f"{name}: correct!")
    else:
        _report(name, False, f"{name}: not quite yet. Your value is {value!r}. {hint}".strip())

def progress():
    passed, total = sum(_results.values()), len(_results)
    print(f"You have passed {passed} of {total} checks so far.")
    if total > 0 and passed == total:
        print("🎉 All checks passed. Well done!")
````

## Part 1 · What is a notebook?
📘 A notebook is a document made of **cells**. There are two kinds:

| Cell type | What it holds | How it looks |
|---|---|---|
| **Text (Markdown) cell** | Explanations, headings, lists (like this one) | Plain text |
| **Code cell** | Python instructions that the computer can run | Grey box, often with `[ ]` or a ▶ button beside it |

**To run a cell:** click on it, then press **Shift + Enter**. The result appears just below the cell.

### Useful shortcuts
| Action | Google Colab | Jupyter Notebook |
|---|---|---|
| Run cell and move to the next | Shift + Enter | Shift + Enter |
| Run cell and stay | Ctrl + Enter | Ctrl + Enter |
| Add a cell below | **+ Code** button | **+** button in the toolbar |
| Undo | Ctrl + Z | Ctrl + Z |
| Start fresh and run everything | Runtime → *Restart session and run all* | Kernel → *Restart & Run All* |

(On a Mac, use **Cmd** instead of **Ctrl**.)

⚠️ **Order matters.** Python remembers what you ran earlier, but only in the order you ran it. If something behaves strangely, restart and run all cells from the top.

### 🔍 Syntax explained
```
print("Hello, health data!")
│    ││                   ││
│    ││                   │└── closing bracket: the end of what we hand to print
│    │└───────────────────┘
│    │        └── the text to show, inside quotes (Python calls text a "string")
│    └── opening bracket: everything inside the brackets is handed to print
└── print: a built-in *function*, a ready-made tool that shows things on the screen
```
- **Quotes** tell Python *"this is text, not an instruction"*. You can use `"double"` or `'single'` quotes, but the opening and closing quote must match.
- **Brackets** `( )` always come as a pair. Forgetting the closing one is the most common beginner error.

### ▶️ Worked example 3: creating variables

## Part 6 · Environment check
📘 The next cell checks that everything is installed. You do **not** need to understand it yet: it uses *loops* and *try/except*, which you will learn on Day 2. Just run it.

You want to see ✅ next to every package and the message **READY FOR DAY 1**.

````python
import sys, platform, importlib

print("Python version:", platform.python_version())
print("Running in Google Colab:", "google.colab" in sys.modules)
print()

all_ok = True
for package in ["numpy", "pandas", "matplotlib", "wbgapi"]:
    try:
        module = importlib.import_module(package)
        print("✅", package, getattr(module, "__version__", ""))
    except ImportError:
        print("❌", package, "is missing. See the troubleshooting table below.")
        all_ok = False

print()
print("READY FOR DAY 1 ✅" if all_ok else "NOT READY YET ❌ (see troubleshooting below)")
````

### 🛠️ Troubleshooting
| What you see | What it means | What to do |
|---|---|---|
| `ModuleNotFoundError` or ❌ next to a package | The package is not installed | Run the Part 0 cell again, then restart the kernel/runtime and run the check again |
| `NameError: name '____' is not defined` | A blank is not filled in yet | Replace `____` with your answer |
| `SyntaxError` | Python cannot understand the line (often a missing quote or bracket) | Look for a missing `"` or `)` on that line |
| Cell shows `[*]` for a long time (Jupyter) | The cell is still running | Wait, or use **Kernel → Interrupt** |
| "Session disconnected" (Colab) | Colab timed out | Click **Reconnect**, then **Runtime → Run all** |
| Shift + Enter does nothing | The cell is not selected | Click inside the cell first |
