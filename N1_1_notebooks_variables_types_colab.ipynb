## Setup
📘 Run the next cell **once**. It creates the *checker* that marks your answers. You do not need to understand it yet.

````python
# 🧰 COURSE HELPER: just run this cell once. (You will understand every line by Day 2.)
_results = {}

def _report(name, ok, message):
    _results[name] = ok
    print(("✅ " if ok else "❌ ") + message)

def check(name, expected, tol=0, hint=""):
    """Is the variable called `name` equal to the expected answer?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. Fill in the blank or write your code, then run the cell again. {hint}".strip())
        return
    value = globals()[name]
    try:
        if isinstance(expected, bool):
            ok = isinstance(value, bool) and value == expected
        elif isinstance(expected, (int, float)):
            ok = (not isinstance(value, bool)) and abs(value - expected) <= tol
        else:
            ok = value == expected
    except Exception:
        ok = False
    if ok:
        _report(name, True, f"{name}: correct!")
    else:
        _report(name, False, f"{name}: not quite yet. Your value is {value!r}. {hint}".strip())

def check_type(name, expected_type, hint=""):
    """Does the variable called `name` hold the expected type of value?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. {hint}".strip())
        return
    value = globals()[name]
    ok = type(value) is expected_type
    if ok:
        _report(name, True, f"{name}: correct type ({expected_type.__name__})!")
    else:
        _report(name, False, f"{name}: expected type {expected_type.__name__}, but yours is {type(value).__name__}. {hint}".strip())

def check_that(name, test, hint=""):
    """Does the value in `name` pass a custom test?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. {hint}".strip())
        return
    value = globals()[name]
    try:
        ok = bool(test(value))
    except Exception:
        ok = False
    if ok:
        _report(name, True, f"{name}: correct!")
    else:
        _report(name, False, f"{name}: not quite yet. Your value is {value!r}. {hint}".strip())

def progress():
    passed, total = sum(_results.values()), len(_results)
    print(f"You have passed {passed} of {total} checks so far.")
    if total > 0 and passed == total:
        print("🎉 All checks passed. Well done!")
````

## Part A · Python as a calculator
📘 Python knows the maths operators you already use, plus a few extras:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | add | `120 + 8` | 128 |
| `-` | subtract | `120 - 80` | 40 |
| `*` | multiply | `18 * 4` | 72 |
| `/` | divide (always gives a decimal) | `70 / 4` | 17.5 |
| `**` | power (exponent) | `1.75 ** 2` | 3.0625 |
| `//` | whole-number division ("how many times does it fit?") | `130 // 60` | 2 |
| `%` | remainder ("what is left over?") | `130 % 60` | 10 |

### ▶️ Worked example A1: several calculations
Remember: when a cell has several lines, use `print( )` to see each result.

````python
print(120 + 8)
print(120 - 80)
print(18 * 4)
print(70 / 4)
print(1.75 ** 2)     # 1.75 squared
print(130 // 60)     # 130 minutes: how many whole hours?
print(130 % 60)      # ... and how many minutes are left over?
````

### ▶️ Worked example A2: order of operations (a classic BMI trap!)
Python follows the same order as maths: **brackets first, then powers, then × and ÷, then + and −**. Compare these two lines carefully:

````python
weight = 70
height = 1.75

print(weight / height ** 2)       # power first, then divide: 22.86 (correct BMI)
print((weight / height) ** 2)     # divide first, then square: 1600 (WRONG!)
print(weight / (height * height)) # brackets make the intention clear: 22.86
````

📘 **Lesson:** when in doubt, use brackets `( )`. They cost nothing and prevent expensive mistakes.

## Part B · Variables: labelled boxes
📘 A **variable** stores a value under a name so you can reuse it. Think of a **labelled box**.

```
weight_kg = 70
│           │
│           └── the value to store
└── the variable name (the label on the box)
    "=" means "store the right-hand value in the left-hand box" (it is NOT the maths "equals")
```

### 📘 Rules for variable names
| Rule | ✅ Good | ❌ Bad |
|---|---|---|
| Letters, digits and underscores `_` only | `blood_pressure` | `blood pressure` (space), `weight-kg` (hyphen) |
| Cannot start with a digit | `reading_2` | `2nd_reading` |
| **Case matters** | `Weight_kg` and `weight_kg` are *different* boxes | |
| Choose clear names | `heart_rate` | `x`, `hr2` |

Course style: **lowercase words joined by underscores** (called *snake_case*), with the **unit** in the name when useful (`weight_kg`, `height_m`).

### ▶️ Worked example B1: create and use variables

````python
weight_kg = 70
height_m = 1.75

print("Weight:", weight_kg, "kg")
print("Height:", height_m, "m")

weight_kg = 72                      # a variable can be changed: the old value is replaced
print("New weight:", weight_kg, "kg")
````

## Part C · Data types
📘 Every value has a **type**. The four you need first:

| Type | Python name | What it holds | Health example |
|---|---|---|---|
| Whole number | `int` | numbers without decimals | `45` (age in years) |
| Decimal number | `float` | numbers with a decimal point | `36.6` (temperature °C) |
| Text | `str` (string) | words inside quotes | `"P001"` (patient ID) |
| True / False | `bool` (boolean) | only `True` or `False` | `False` (is a smoker?) |

Why do types matter? `"70"` (text) and `70` (number) look alike but behave differently: you can do maths with the number, not with the text.

### 📘 Functions in one minute
`type(45)` is a **function call**: a function is a ready-made mini-machine. You put something **into the brackets**, and it gives something back.
```
type(45)
│    └── the input (called an "argument")
└── the function name
```

### ▶️ Worked example C1: asking Python for the type

````python
print(type(45))         # int
print(type(36.6))       # float
print(type("P001"))     # str
print(type(True))       # bool
print(type("70"))       # str: text, even though it looks like a number!
````

### ▶️ Worked example C2: converting between types
Lab files often deliver numbers **as text**. Three converting functions do the job: `int( )`, `float( )`, `str( )`.

````python
age_text = "45"
age_number = int(age_text)        # text -> whole number
print(age_number + 1)             # now we can do maths: 46

temperature = 36.6
print("Temperature: " + str(temperature) + " °C")   # number -> text so it can be joined to text
````

### ▶️ Worked example C3: comparisons give True or False
`>` greater than · `<` less than · `>=` at least · `<=` at most · `==` **is equal to** (two equals signs!) · `!=` is not equal to

````python
age = 45
print(age > 60)       # False
print(age >= 45)      # True
print(age == 45)      # True: "is age equal to 45?"
print(age != 45)      # False
````

## Part D · Worked case: Body Mass Index (BMI)
📘 **BMI = weight (kg) ÷ height (m)²**. It is a quick screening measure used in public health. (Educational example: not a diagnosis.)

### ▶️ Worked example D1: BMI for one patient, step by step

````python
# Step 1: store the measurements (include the unit in the name!)
weight_kg = 70
height_m = 1.75

# Step 2: apply the formula
bmi = weight_kg / (height_m ** 2)

# Step 3: show the result. round(number, 1) keeps 1 decimal place
print("BMI:", bmi)
print("BMI (rounded):", round(bmi, 1))
````

## Part E · Reading error messages
📘 Every programmer sees errors every day. An error is Python telling you **exactly** what it cannot do. **Read the last line first.**

A typical error looks like this:
```
Traceback (most recent call last):
  File "<cell>", line 2, in <module>
    message = "Weight: " + weight_kg
TypeError: can only concatenate str (not "int") to str
```
- **Last line** = the error type (`TypeError`) and its explanation. Start here.
- **Line above** = the exact line that failed.

### The four errors you will meet most often
| Error | Usually means | Fix |
|---|---|---|
| `NameError` | you used a name that does not exist (typo, or you forgot to run an earlier cell) | check spelling; run earlier cells |
| `TypeError` | you mixed types that do not go together (text + number) | convert with `str( )`, `int( )` or `float( )` |
| `SyntaxError` | Python cannot read the line (missing quote or bracket) | look for a missing `"` or `)` |
| `ZeroDivisionError` | you divided by zero | check the divisor |

### ▶️ Worked example E1: seeing errors safely
The cell below causes three errors **on purpose**. We wrap each in `try / except` so the notebook keeps going and prints Python's message. (You will learn `try / except` on Day 2. For now, just read the messages.)

````python
try:
    print(wieght_kg)                # typo in the variable name
except NameError as error:
    print("NameError:", error)

try:
    weight_kg = 70
    print("Weight: " + weight_kg)    # text + number
except TypeError as error:
    print("TypeError:", error)

try:
    print(10 / 0)                   # division by zero
except ZeroDivisionError as error:
    print("ZeroDivisionError:", error)
````

## Part F · Your turn: Body Surface Area (BSA)
📘 **BSA** is used to dose some medicines (for example in chemotherapy). The **Mosteller formula** is:

**BSA (m²) = square root of [ height (cm) × weight (kg) ÷ 3600 ]**

Python has no square-root symbol at this stage, but a square root is the same as raising to the power **0.5**: `x ** 0.5`.

### ▶️ Worked example F1: a patient of 170 cm and 70 kg

````python
height_cm = 170
weight_kg = 70
bsa = (height_cm * weight_kg / 3600) ** 0.5
print("BSA:", round(bsa, 2), "m²")      # 1.82
````
