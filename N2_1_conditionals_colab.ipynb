## Setup

````python
# 🧰 COURSE HELPER: just run this cell once. (You will understand every line by Day 2.)
_results = {}

def _report(name, ok, message):
    _results[name] = ok
    print(("✅ " if ok else "❌ ") + message)

def check(name, expected, tol=0, hint=""):
    """Is the variable called `name` equal to the expected answer?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. Fill in the blank or write your code, then run the cell again. {hint}".strip())
        return
    value = globals()[name]
    try:
        if isinstance(expected, bool):
            ok = isinstance(value, bool) and value == expected
        elif isinstance(expected, (int, float)):
            ok = (not isinstance(value, bool)) and abs(value - expected) <= tol
        else:
            ok = value == expected
    except Exception:
        ok = False
    if ok:
        _report(name, True, f"{name}: correct!")
    else:
        _report(name, False, f"{name}: not quite yet. Your value is {value!r}. {hint}".strip())

def check_type(name, expected_type, hint=""):
    """Does the variable called `name` hold the expected type of value?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. {hint}".strip())
        return
    value = globals()[name]
    ok = type(value) is expected_type
    if ok:
        _report(name, True, f"{name}: correct type ({expected_type.__name__})!")
    else:
        _report(name, False, f"{name}: expected type {expected_type.__name__}, but yours is {type(value).__name__}. {hint}".strip())

def check_that(name, test, hint=""):
    """Does the value in `name` pass a custom test?"""
    if name not in globals():
        _report(name, False, f"{name}: I can't find this variable yet. {hint}".strip())
        return
    value = globals()[name]
    try:
        ok = bool(test(value))
    except Exception:
        ok = False
    if ok:
        _report(name, True, f"{name}: correct!")
    else:
        _report(name, False, f"{name}: not quite yet. Your value is {value!r}. {hint}".strip())

def progress():
    passed, total = sum(_results.values()), len(_results)
    print(f"You have passed {passed} of {total} checks so far.")
    if total > 0 and passed == total:
        print("🎉 All checks passed. Well done!")


def check_function(name, cases, tol=0, hint=""):
    """Call the function called `name` with test inputs and compare the results with the expected outputs."""
    if name not in globals() or not callable(globals()[name]):
        _report(name, False, f"{name}: I can't find a function with this name yet. {hint}".strip())
        return
    func = globals()[name]
    for args, expected in cases:
        try:
            result = func(*args)
            if isinstance(expected, bool):
                ok = isinstance(result, bool) and result == expected
            elif isinstance(expected, (int, float)):
                ok = (not isinstance(result, bool)) and result is not None and abs(result - expected) <= tol
            else:
                ok = result == expected
        except Exception:
            ok, result = False, "(an error)"
        if not ok:
            _report(name, False, f"{name}{args}: your function gave {result!r}, which is not what I expected. {hint}".strip())
            return
    _report(name, True, f"{name}: passed all {len(cases)} test cases!")
````

## Part A · `if` and `else`: choosing what to do
📘 An **`if` statement** runs some lines **only when a condition is True**.
```
if glucose_mgdl >= 126:
│  │                  │
│  │                  └── a COLON ":" at the end means "then do the following"
│  └── the condition: it becomes True or False
└── the keyword  if
    print("High glucose")      ← indented by 4 spaces: these lines BELONG to the if
print("Done")                  ← NOT indented: this line always runs
```
### The two rules you must remember
1. **Colon `:`** at the end of every `if`, `elif` and `else` line.
2. **Indentation:** lines that belong to the `if` are pushed in by **4 spaces** (the **Tab** key does this in a notebook). The block ends when the indentation goes back.

`else:` gives the plan B: it runs when the condition is **False**.

### ▶️ Worked example A1: if / else

````python
glucose_mgdl = 135

if glucose_mgdl >= 126:
    print("Glucose is in the diabetes range.")
    print("Suggest a repeat test.")
else:
    print("Glucose is below the diabetes range.")

print("Assessment finished.")        # not indented: runs in every case
````

## Part B · `elif` chains: more than two outcomes
📘 `elif` (short for *else if*) lets you test several conditions **in order**. Python runs **only the first branch whose condition is True**, then skips the rest.

| Category (WHO adult BMI) | Condition |
|---|---|
| Underweight | BMI < 18.5 |
| Normal weight | 18.5 ≤ BMI < 25 |
| Overweight | 25 ≤ BMI < 30 |
| Obesity | BMI ≥ 30 |

Notice that each condition only needs the **upper limit**: by the time Python reaches `elif bmi < 25`, it already knows BMI is not below 18.5.

### ▶️ Worked example B1: an elif chain

````python
bmi = 27.4

if bmi < 18.5:
    category = "Underweight"
elif bmi < 25:
    category = "Normal weight"
elif bmi < 30:
    category = "Overweight"
else:
    category = "Obesity"

print(f"BMI {bmi} -> {category}")      # BMI 27.4 -> Overweight
````

📘 **Try it:** change `bmi` to `17`, `22`, `31`, and exactly `25`. Notice that `25` lands in *Overweight*: the condition is `bmi < 25`, so 25 itself is **not** below 25. **Boundary values are where bugs hide.**

## Part C · `and`, `or`, `not`: combining conditions
📘 Real rules usually combine several tests.

| Word | Meaning | Example | True when… |
|---|---|---|---|
| `and` | **both** must be True | `age >= 40 and age <= 74` | the age is between 40 and 74 |
| `or` | **at least one** must be True | `systolic >= 140 or diastolic >= 90` | either reading is high |
| `not` | flips True ↔ False | `not has_diagnosis` | there is no diagnosis |

**Truth table** (T = True, F = False):

| A | B | `A and B` | `A or B` |
|---|---|---|---|
| T | T | T | T |
| T | F | F | T |
| F | T | F | T |
| F | F | F | F |

### ▶️ Worked example C1: blood pressure categories
Teaching scheme (simplified; use the guideline your clinic follows):

| Category | Rule |
|---|---|
| Hypertension range | systolic ≥ 140 **or** diastolic ≥ 90 |
| Elevated | systolic ≥ 120 **or** diastolic ≥ 70 (and not in the hypertension range) |
| Non-elevated | everything else |

Because the **first** matching branch wins, we test the most severe category **first**.

````python
systolic = 135
diastolic = 72

if systolic >= 140 or diastolic >= 90:
    bp_example = "Hypertension range"
elif systolic >= 120 or diastolic >= 70:
    bp_example = "Elevated"
else:
    bp_example = "Non-elevated"

print(f"{systolic}/{diastolic} -> {bp_example}")      # 135/72 -> Elevated
````

## Part D · Mini-project: screening flags
📘 For each patient we will decide: BMI category, glucose category, and whether the patient needs **follow-up** (rule: *glucose in the diabetes range* **or** *BMI of 30 or more*).

The three patients are the same ones as in Day 1.

### ▶️ Worked example D1: patient P001 (index 0)

````python
patients = [
    {"id": "P001", "age": 54, "sex": "F", "weight_kg": 68.0, "height_cm": 165, "glucose_mgdl": 126, "creatinine_umol_l": 88},
    {"id": "P002", "age": 61, "sex": "M", "weight_kg": 82.5, "height_cm": 178, "glucose_mgdl": 98,  "creatinine_umol_l": 110},
    {"id": "P003", "age": 47, "sex": "F", "weight_kg": 59.0, "height_cm": 160, "glucose_mgdl": 182, "creatinine_umol_l": 70},
]

p = patients[0]
bmi = p["weight_kg"] / (p["height_cm"] / 100) ** 2

if bmi < 18.5:
    bmi_cat_p1 = "Underweight"
elif bmi < 25:
    bmi_cat_p1 = "Normal weight"
elif bmi < 30:
    bmi_cat_p1 = "Overweight"
else:
    bmi_cat_p1 = "Obesity"

if p["glucose_mgdl"] < 100:
    glucose_cat_p1 = "Normal"
elif p["glucose_mgdl"] < 126:
    glucose_cat_p1 = "Prediabetes"
else:
    glucose_cat_p1 = "Diabetes range"

follow_up_p1 = p["glucose_mgdl"] >= 126 or bmi >= 30

print(p["id"], "|", bmi_cat_p1, "|", glucose_cat_p1, "| follow-up:", follow_up_p1)
````

## 🧾 Recap: your cheat sheet
| Idea | Syntax | Example |
|---|---|---|
| Decide | `if condition:` | `if glucose >= 126:` |
| Plan B | `else:` | |
| Several options | `elif condition:` | `elif bmi < 30:` |
| Both true | `and` | `age >= 40 and age <= 74` |
| Either true | `or` | `systolic >= 140 or diastolic >= 90` |
| Flip | `not` | `not has_diagnosis` |
| Store a True/False | `x = a > b` | `eligible = age >= 40` |

### Common slips
- **Forgetting the colon** `:` → `SyntaxError`.
- **Wrong indentation** → `IndentationError` (lines in the same block must line up; use 4 spaces or the Tab key).
- Using `=` (store) instead of `==` (compare) inside an `if`.
- **Wrong order:** put the **most severe or most specific** condition first.
- **Boundary values:** check what happens at exactly 25, 30, 126…
